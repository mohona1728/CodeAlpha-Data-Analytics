```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

print("Libraries imported successfully!")
```

```python
# Step 2: Scrape book information from a demo website

url = "https://books.toscrape.com/"

response = requests.get(url, timeout=30)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

books = []

for book in soup.select("article.product_pod"):
    title = book.select_one("h3 a")["title"]
    price = book.select_one("p.price_color").get_text(strip=True)
    availability = book.select_one("p.availability").get_text(strip=True)

    rating_class = book.select_one("p.star-rating")["class"][1]

    rating_map = {
        "One": 1,
        "Two": 2,
        "Three": 3,
        "Four": 4,
        "Five": 5
    }

    rating = rating_map[rating_class]

    books.append({
        "Title": title,
        "Price": price,
        "Availability": availability,
        "Rating": rating
    })

books_df = pd.DataFrame(books)

print("Total books scraped:", len(books_df))
display(books_df.head(10))
```

```python
# Step 3: Save scraped data as a CSV file

csv_filename = "CodeAlpha_Task1_Scraped_Books.csv"

books_df.to_csv(csv_filename, index=False, encoding="utf-8-sig")

print("CSV file saved successfully:", csv_filename)
```
