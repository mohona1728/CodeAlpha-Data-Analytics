```python
# Task 4: Sentiment Analysis
# Step 1: Install and import the required library

!pip -q install vaderSentiment

import pandas as pd
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

print("Libraries imported successfully!")
```

```python
# Step 2: Create a sample customer review dataset

data = {
    "Review_ID": range(1, 16),

    "Customer_Review": [
        "This product is absolutely amazing. I love it!",
        "The quality is excellent and delivery was fast.",
        "Very happy with my purchase.",
        "Great service and wonderful experience.",
        "I am satisfied with this product.",
        "The product is okay, nothing special.",
        "Average quality and average service.",
        "It is neither good nor bad.",
        "The product is acceptable.",
        "Not bad, but could be better.",
        "Very poor quality. I am disappointed.",
        "Terrible experience. I hate this product.",
        "Delivery was late and customer service was poor.",
        "The product stopped working after one day.",
        "I am extremely unhappy with this purchase."
    ]
}

reviews_df = pd.DataFrame(data)

print("Customer review dataset created successfully!")
display(reviews_df)
```

```python
# Step 3: Perform Sentiment Analysis

def analyze_sentiment(review):
    score = analyzer.polarity_scores(review)["compound"]

    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"

reviews_df["Sentiment"] = reviews_df["Customer_Review"].apply(
    analyze_sentiment
)

display(reviews_df)
```

```python
# Step 4: Count each sentiment category

sentiment_counts = reviews_df["Sentiment"].value_counts()

print("Sentiment Analysis Summary:")
display(sentiment_counts)

print("\nPercentage of each sentiment:")
sentiment_percentage = reviews_df["Sentiment"].value_counts(normalize=True) * 100
display(sentiment_percentage.round(2))
```

```python
# Step 5: Visualize Sentiment Analysis Results

import matplotlib.pyplot as plt

# Bar Chart
plt.figure(figsize=(8, 5))
sentiment_counts = reviews_df["Sentiment"].value_counts()

sentiment_counts.plot(kind="bar")

plt.title("Customer Review Sentiment Analysis")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()


# Pie Chart
plt.figure(figsize=(7, 7))

plt.pie(
    sentiment_counts,
    labels=sentiment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sentiment Distribution")
plt.show()
```

```python
# Step 6: Export the Sentiment Analysis Dataset

reviews_df.to_csv(
    "CodeAlpha_Task4_Sentiment_Analysis.csv",
    index=False
)

print("Sentiment Analysis dataset exported successfully!")
```
