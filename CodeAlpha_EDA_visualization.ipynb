```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Department": [
        "Medicine", "Surgery", "Medicine", "Pediatrics",
        "Surgery", "Medicine", "Pediatrics", "Surgery",
        "Medicine", "Pediatrics"
    ],
    "Waiting_Minutes": [20, 35, 15, 25, 40, 30, 10, 45, 25, 20],
    "Satisfaction": [4, 3, 5, 4, 2, 4, 5, 2, 4, 5]
}

df = pd.DataFrame(data)

print("First five rows:")
display(df.head())

print("Dataset information:")
df.info()

print("Statistical summary:")
display(df.describe())

print("Missing values:")
print(df.isnull().sum())

print("Department counts:")
print(df["Department"].value_counts())
```

Key Findings

1. Dataset Overview

The dataset contains 10 records and 3 variables: Department, Waiting_Minutes, and Satisfaction. It includes data from three departments: Medicine, Surgery, and Pediatrics.

2. Average Waiting Time

The overall average waiting time was 26.5 minutes. The longest waiting time was 45 minutes, while the shortest waiting time was 10 minutes.

3. Department-wise Waiting Time

The Surgery department had the highest average waiting time of 40 minutes, followed by Medicine at 22.5 minutes and Pediatrics at approximately 18.33 minutes.

4. Patient Satisfaction

The average patient satisfaction score was 3.8 out of 5. The scores ranged from 2 to 5, indicating variation in reported satisfaction levels.

5. Data Quality Assessment

The dataset contained no missing values in any of the three variables. All 10 records had complete values for Department, Waiting_Minutes, and Satisfaction.

6. Overall Interpretation

The analysis showed that waiting times varied across departments, with Surgery having the highest average waiting time. The overall average satisfaction score was 3.8 out of 5. These findings can help identify areas for further investigation and support discussions about healthcare service efficiency and patient experience.

7. Limitation

This analysis is based on a small dataset of only 10 records. Therefore, the findings should not be generalized to all hospital patients or departments. Further analysis using a larger, representative dataset would be needed to draw stronger conclusions.

```python
print("Average waiting time:",
      df["Waiting_Minutes"].mean())

print("Longest waiting time:",
      df["Waiting_Minutes"].max())

print("Shortest waiting time:",
      df["Waiting_Minutes"].min())

print("Average satisfaction:",
      df["Satisfaction"].mean())

print("Average waiting time by department:")
display(
    df.groupby("Department")["Waiting_Minutes"]
    .mean()
    .sort_values(ascending=False)
)
```

```python
import matplotlib.pyplot as plt

# Average waiting time by department
department_wait = df.groupby("Department")["Waiting_Minutes"].mean()

plt.figure(figsize=(8, 5))
department_wait.plot(kind="bar")
plt.title("Average Waiting Time by Department")
plt.xlabel("Department")
plt.ylabel("Average Waiting Time (Minutes)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Calculate average patient satisfaction
avg_satisfaction = df["Satisfaction"].mean()

print("Average Patient Satisfaction:",
      round(avg_satisfaction, 2), "out of 5")

# Chart 1: Satisfaction score distribution
plt.figure(figsize=(7, 5))
df["Satisfaction"].value_counts().sort_index().plot(
    kind="bar"
)
plt.title("Distribution of Patient Satisfaction Scores")
plt.xlabel("Satisfaction Score (1-5)")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Chart 2: Average satisfaction by department
dept_satisfaction = df.groupby("Department")["Satisfaction"].mean()

plt.figure(figsize=(8, 5))
dept_satisfaction.plot(kind="bar")
plt.title("Average Patient Satisfaction by Department")
plt.xlabel("Department")
plt.ylabel("Average Satisfaction Score")
plt.ylim(0, 5)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("\nAverage satisfaction by department:")
print(dept_satisfaction.round(2))
```

Patient Satisfaction Analysis

The overall average patient satisfaction score was 3.8 out of 5. The satisfaction scores ranged from 2 to 5. The score distribution chart shows how frequently each satisfaction rating occurred, while the department-wise chart compares average satisfaction across Medicine, Surgery, and Pediatrics. These visualizations help summarize patient satisfaction patterns in the dataset. However, the small sample size of 10 records limits the generalizability of the findings.
